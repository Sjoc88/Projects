```python
import pandas as pd

# 📂 Cargar el archivo CSV generado anteriormente
df = pd.read_csv("datos_xml_con_fechas.csv")

# 📌 1. Eliminar las columnas innecesarias (A, B y E)
df = df.drop(columns=["Archivo", "IdentificacionSeriesTemporales", "UnidadPrecio"], errors="ignore")

# 📌 2. Filtrar filas donde:
# - 'TerminoCosteHorario' sea "FEU"
# - 'TipoPrecio' sea "Z14"
df = df[(df["TerminoCosteHorario"] == "FEU") & (df["TipoPrecio"] == "Z14")]

# 📌 3. Renombrar la columna 'POS' a 'Hour' y reemplazar 24 con 0
df = df.rename(columns={"Pos": "Hour"})
df["Hour"] = df["Hour"].replace(24, 0)

# 📌 4. Ordenar el DataFrame por 'Fecha' y 'Hour'
df = df.sort_values(by=["Fecha", "Hour"], ascending=[True, True])

# 📌 5. Guardar el archivo final limpio
df.to_csv("datos_xml_limpios.csv", index=False, encoding="utf-8")

# 📊 Mostrar las primeras filas del resultado
print(df.head())
```
