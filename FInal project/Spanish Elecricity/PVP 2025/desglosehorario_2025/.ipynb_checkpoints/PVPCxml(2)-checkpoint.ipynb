```python
import pandas as pd

# 📂 Cargar el archivo CSV generado antes
df = pd.read_csv("datos_xml.csv")  # Asegúrate de que el archivo tiene la primera columna con el nombre del XML

# 📌 Extraer la fecha desde el nombre del archivo
df["Fecha"] = df["Archivo"].str[-12:-4]  # Extrae 'yyyymmdd' del nombre del archivo

# 📆 Convertir la fecha al formato mm/dd/yyyy
df["Fecha"] = pd.to_datetime(df["Fecha"], format="%Y%m%d").dt.strftime("%m/%d/%Y")

# 📝 Guardar el DataFrame con la fecha corregida
df.to_csv("datos_xml_con_fechas.csv", index=False)

# 📊 Mostrar las primeras filas
print(df.head())
```
