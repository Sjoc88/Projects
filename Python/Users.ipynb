```python
# Importamos las librerías que necesitamos
import pandas as pd
import numpy as np

user = pd.read_csv('DatosUsuariosRedesSociales.csv')

user.sample(5)
```

```python
user.groupby('platform').mean(numeric_only = True)
```

```python
edad_media = user.groupby('platform')['age'].mean()
edad_media
```

```python
tiempo_uso = user.groupby('location')['time_spent'].sum().reset_index()
tiempo_uso
```

```python
salario_profesion = user.groupby('profession')['income'].agg(['mean','median','std','max','min']).reset_index()
salario_profesion
```

```python
def categorizar_edad (numero):
    if 18 <= numero <= 28:
        return 'Jóven adulto'
    elif 29 <= numero <= 39:
        return 'Adulto jóven'
    elif 40 <= numero <= 50:
        return 'Mediana edad'
    else:
        return 'Adultos mayores'
```

```python
user['age_category'] = user['age'].apply(categorizar_edad)
user
```

```python
user['income'].dtype
```

```python
def cambiar_float(numero):
    return float(numero)

user['income'] = user['income'].apply(cambiar_float)

user['income'].dtype
```

```python
user['time_spent'] = user['time_spent'].apply(lambda x: float(x))

user['time_spent'].dtype
```

```python
def travel_female(col_interests,col_gender):

    if col_interests == 'Travel' and col_gender == 'female':
        return True
    else:
        return False
```

```python
user['target_audiences']=user.apply(lambda x: travel_female(x['interests'],x['gender']),axis=1)
user
```
