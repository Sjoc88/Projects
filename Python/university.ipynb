```python
import pandas as pd
```

```python
input_file = "DatosPandasUniversidades.csv"
```

```python
df = pd.read_csv(input_file, index_col=2)
```

```python
df.head()
```

```python
condición_USA = df["country"] == "United States"
universidades_USA = df[condición_USA]
universidades_USA.head()
```

```python
condición_90 = df["overall_score"] > 90.0
universidades_90 = df[condición_90]
universidades_90.head()
```

```python
universidades_USA_90 = df[condición_90 & condición_USA]
universidades_USA_90.head()
```

```python
#Vamos a crear una columna con la media de la reputación académica y de empleado
calculo_media = (df['academic_reputation'] + df['employer_reputation'])/2
df['mean_reputation_direct_assignment'] = calculo_media
```
