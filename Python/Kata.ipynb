```python
# 1. Escribe una función que reciba una cadena de texto como parámetro y devuelva un diccionario con las frecuencias de cada letra en la cadena. Los espacios no deben ser considerados

# Entrada: una cadena de texto (string)
# Salida: un diccionario
# Clave: cada letra
# Valor: cuántas veces aparece
# : los espacios " " NO 

def frecuencia_letras(cadena):
    cadena_limpia = cadena.replace(" ", "").lower()
    frecuencia = {}
    for letra in cadena_limpia:
        if letra != " ":  # Ignorar espacios
            if letra in frecuencia:
                frecuencia[letra] += 1
            else:
                frecuencia[letra] = 1
    return frecuencia
```

```python
resultado = frecuencia_letras("Hola, me llamo Stéphane")
print(resultado)
```

```python
# 3. Escribe una función que tome una lista de palabras y una palabra objetivo como parámetros. La función debe
#  devolver una lista con todas las palabras de la lista original que contengan la palabra objetivo.

def filtrar_palabras(lista_palabras, objetivo):
    resultados = []
    for palabra in lista_palabras:
        if objetivo in palabra:
            resultados.append(palabra)
    return resultados
```

```python
palabras = ["casa", "casamiento", "perro", "mi casa es tu casa", "gato"]
objetivo = "casa"
print(filtrar_palabras(palabras, objetivo))
```

```python
# 5. Escribe una función que tome una lista de números como parámetro y un valor opcional nota_aprobado, que por 
# defecto es 5. La función debe calcular la media de los números en la lista y determinar si la media es mayor o igual 
# que nota aprobado. Si es así, el estado será "aprobado", de lo contrario, será "suspenso". La función debe devolver
# una tupla que contenga la media y el estado.

def evaluar_nota(numeros, nota_aprobado=5):
    media = sum(numeros) / len(numeros)
    estado = "aprobado" if media >= nota_aprobado else "suspenso"
    return (media, estado)

notas = [6, 7, 8, 4, 5]
resultado_evaluacion = evaluar_nota(notas)
print(resultado_evaluacion)
```

```python
# 6. Escribe una función que calcule el factorial de un número de manera recursiva.

def factorial(n):
    if n == 0:
        return 1
    elif n == 1:
        return 1
    else:
        return n * factorial(n - 1)
    
numero = 5
print(factorial(numero))


# Tambien podemos pedir al usuario que ingrese un numero para calcular su factorial

numero = int(input("Introduce un número para calcular su factorial: "))
print(f"El factorial del numero introducido {numero} es {factorial(numero)}")
```

```python
# 8. Escribe un programa que pida al usuario dos números e intente dividirlos. Si el usuario ingresa un valor no numérico 
# o intenta dividir por cero, maneja esas excepciones de manera adecuada. Asegúrate de mostrar un mensaje indicando si la 
# división fue exitosa o no.

def dividir_numeros():
    try:
        num1 = float(input("Introduce el primer número: "))
        num2 = float(input("Introduce el segundo número: "))
        resultado = num1 / num2
    except ValueError:
        print("Error: Debes ingresar valores numéricos.")
    except ZeroDivisionError:
        print("Error: No se puede dividir por cero.")
    else:
        print(f"La división fue exitosa. El resultado es: {resultado}")

print(dividir_numeros())
```

```python
# 9. Escribe una función que tome una lista de nombres de mascotas como parámetro y devuelva una nueva lista excluyendo ciertas 
# mascotas prohibidas en España. La lista de mascotas a excluir es ["Mapache", "Tigre", "Serpiente Pitón", "Cocodrilo", "Oso"].
# Usa la función filter()

def filtrar_mascotas(lista_mascotas):
    mascotas_prohibidas = ["Mapache", "Tigre", "Serpiente Pitón", "Cocodrilo", "Oso"]
    def no_prohibida(mascota):
        return mascota not in mascotas_prohibidas
    return list(filter(no_prohibida, lista_mascotas))
mascotas = ["Perro", "Gato", "Mapache", "Tigre", "Loro", "Cocodrilo"]
mascotas_permitidas = filtrar_mascotas(mascotas)
print(mascotas_permitidas)
```

```python
# 10. Escribe una función que reciba una lista de números y calcule su promedio. Si la lista está vacía, lanza una excepción 
# personalizada y maneja el error adecuadamente.

class ListaVaciaError(Exception):
    pass
def calcular_promedio(numeros):
    if not numeros:
        raise ListaVaciaError("La lista está vacía. No se puede calcular el promedio.")
    return sum(numeros) / len(numeros)
try:
    lista_numeros = []
    promedio = calcular_promedio(lista_numeros)
    print(f"El promedio es: {promedio}")
except ListaVaciaError as e:
    print(f"Error: {e}")
lista_numeros = [10, 20, 30]
promedio = calcular_promedio(lista_numeros)
print(f"El promedio es: {promedio}")
```

```python
# 13. Genera una función la cual, para un conjunto de caracteres, devuelva una lista de tuplas con cada letra en mayúsculas 
# y minúsculas. Las letras no pueden estar repetidas .Usa la función map()

def letras_mayus_minus(conjunto):
    conjunto_unico = set(conjunto)
    def mayus_minus(letra):
        return (letra.upper(), letra.lower())
    return list(map(mayus_minus, conjunto_unico))

conjunto_caracteres = "abcdeabc"
resultado_letras = letras_mayus_minus(conjunto_caracteres)
print(resultado_letras)
```

```python
# 14. Crea una función que retorne las palabras de una lista de palabras que comience con una letra en especifico. 
# Usa la función filter()

def filtrar_por_letra(lista_palabras, letra):
    def comienza_con(palabra):
        return palabra.startswith(letra)
    return list(filter(comienza_con, lista_palabras))

palabras = ["manzana", "banana", "mango", "pera", "melocotón"]
letra = "m"
resultado_filtrado = filtrar_por_letra(palabras, letra)
print(resultado_filtrado)
```

```python
# 16. Escribe una función que tome una cadena de texto y un número entero n como parámetros y devuelva una lista de todas las 
# palabras que sean más largas que n. Usa la función filter ()

def palabras_mas_largas(cadena, n):
    palabras = cadena.split()
    def es_mas_larga(palabra):
        return len(palabra) > n
    return list(filter(es_mas_larga, palabras))

cadena = "Este es un ejemplo de cadena de texto para filtrar palabras"
n = 3

resultado_palabras = palabras_mas_largas(cadena, n)
print(resultado_palabras)
```

```python
# 28. Crea una función que busque y devuelva el primer elemento duplicado en una lista dada

def primer_duplicado(lista):
    vistos = set()
    for elemento in lista:
        if elemento in vistos:
            return elemento
        vistos.add(elemento)
    return None

lista = [1, 2, 3, 4, 2, 5, 3]
duplicado = primer_duplicado(lista)
print(duplicado)
```

```python
# 29. Crea una función que convierta una variable en una cadena de texto y enmascare todos los caracteres con el carácter '#', 
# excepto los últimos cuatro

def enmascarar_variable(variable):
    cadena = str(variable)
    longitud = len(cadena)
    if longitud <= 4:
        return cadena
    else:
        return '#' * (longitud - 4) + cadena[-4:]
    
variable = 1234567890
resultado_enmascarado = enmascarar_variable(variable)
print(resultado_enmascarado)
```

```python
# 30. Crea una función que determine si dos palabras son anagramas, es decir, si están formadas por las mismas letras pero 
# en diferente orden

def son_anagramas(palabra1, palabra2):
    return sorted(palabra1.lower()) == sorted(palabra2.lower())

palabra1 = "listen"
palabra2 = "silent"
resultado_anagramas = son_anagramas(palabra1, palabra2)
print(resultado_anagramas)
```
